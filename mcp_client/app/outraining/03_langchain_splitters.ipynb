#  LangChain Text Splitters – Concepts and Explanation

This notebook demonstrates how to use different **LangChain text splitters** with non-technical data.

###  What are Text Splitters?

Text splitters divide large documents or data into manageable **chunks**. These chunks are useful for embedding, retrieval, or summarization tasks in LLM pipelines.

---

###  Splitters Used in This Notebook:

- **CharacterTextSplitter**: Splits text by a specified character (e.g., `\n`) using a fixed chunk size and overlap.
- **RecursiveCharacterTextSplitter**: Smart splitter that tries multiple separators in a prioritized order to avoid cutting sentences or words awkwardly.
- **HTMLHeaderTextSplitter**: Parses HTML content and splits it based on heading tags (e.g., `<h1>`, `<h2>`), keeping semantic structure.
- **RecursiveJsonSplitter**: Breaks down deeply nested JSON content into digestible chunks.

---

###  What is RecursiveCharacterTextSplitter?

This splitter tries different splitting strategies recursively. It starts from bigger structural splits (like paragraphs) and goes down to smaller ones (like sentences or words), maintaining:

- Logical completeness
- Text boundaries
- Chunk size constraints

---

###  Notes on Chunking Parameters

| Parameter       | Description |
|----------------|-------------|
| `chunk_size`   | Max number of characters per chunk |
| `chunk_overlap`| Number of characters repeated from the previous chunk for continuity/context |

This ensures **context preservation** across chunks — essential for retrieval-augmented generation (RAG) and summarization.

### Load data files

```python

base_dir = Path("langchain_text_splitter_data")  # Replace with your path
text_path = base_dir / "wild_horses_story.txt"
html_path = base_dir / "historical_events.html"
```

```python
text_content = text_path.read_text()
html_content = html_path.read_text()
json_content=requests.get("https://api.smith.langchain.com/openapi.json").json()
```

### 1. CharacterTextSplitter


```python
# Split text using newline with fixed chunk size and overlap
char_splitter = CharacterTextSplitter(separator="\n", chunk_size=300, chunk_overlap=50)
char_chunks = char_splitter.split_text(text_content)

print("=== CharacterTextSplitter ===")
for i, chunk in enumerate(char_chunks[:2], 1):
# Display chunk samples
    print(f"\nChunk {i}:\n{chunk[:300]}...\n")
```

### 2. RecursiveCharacterTextSplitter

```python
# Split text using newline with fixed chunk size and overlap
recursive_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)
recursive_chunks = recursive_splitter.split_text(text_content)

print("=== RecursiveCharacterTextSplitter ===")
for i, chunk in enumerate(recursive_chunks[:2], 1):
# Display chunk samples
    print(f"\nChunk {i}:\n{chunk[:300]}...\n")
```

### 3. HTMLHeaderTextSplitter


```python
# Split HTML content by <h1> tags
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=[("h1", "section")])
html_chunks = html_splitter.split_text(html_content)


print("=== HTMLHeaderTextSplitter ===")
for i, chunk in enumerate(html_chunks[:2], 1):
# Display chunk samples
    print(f"\nChunk {i}:\n{chunk.page_content[:300]}...\n")
```
